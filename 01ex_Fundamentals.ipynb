4\. Write the factorial function so that it a) does and b) does not use recursion.

````python
from functools import reduce
n = 7

# a) With recursion
def fact1(x):
    if x == 0 or x == 1:
        return 1
    else:
        return x * fact1(x-1)

# b) Whithout recursion
def fact2(x):
    def mult(a,b):
        return a*b
    
    if x == 0 or x == 1:
        return 1
    else:
        return reduce(mult, [i for i in range(1,x+1)])

def fact3(x):
    b = 1
    if x == 0 or x == 1:
        return 1
    for i in range(x):
        b *= (i+1)
    return b


print("Recur",fact1(n))
print("No recur",fact2(n))
print("No recur 2",fact3(n))
````

5\. Use HOFs (zip in particular) to compute the weight of a circle, a disk and a sphere, assuming different radii and different densities:

```python
densities = {"Al":[0.5,1,2],"Fe":[3,4,5],"Pb": [15,20,30]}
radii = [1,2,3]
```

where the entries of the dictionary's values are the linear, superficial and volumetric densities of the materials respectively.

In particular define a list of three lambda functions using a comprehension that computes the circumference, the area and the volume for a given radius.


````python
from math import pi

# Weight = m*g   -->  W = density * V * g
densities = {"Al":[0.1,1,2],"Fe":[3,4,5],"Pb":[15,20,30]}
radii = [1,2,3]
fig = ["circle","disk","sphere"]
g = 9.81

vols = [lambda r: 2*pi*r, lambda r: pi*r*r, lambda r: 4/3 * pi * r*r*r]

for i in densities:
    print("\nMaterial", i)
    for rho, v, f, r in zip(densities[i],vols,fig,radii):
        w = rho * v(r) * g
        print("Weight of",i,f,"is",f"{w:.2f}","N",)
````

````python
# Parent class
class Dog:

    cont = 0
    all_dogs = []

    # Class attribute
    species = 'mammal'

    # Initializer / Instance attributes
    def __init__(self, name, age, is_hungry):
        self.name = name
        self.age = age
        self.is_hungry = is_hungry
        Dog.cont += 1
        Dog.all_dogs.append(self)

    # instance method
    def description(self):
        return "{} is {}.".format(self.name, self.age)

    # instance method
    def speak(self, sound):
        return "{} says {}".format(self.name, sound)

    # instace method
    def eat(self):
        self.is_hungry = False

# Child class (inherits from Dog class)
class RussellTerrier(Dog):
    def run(self, speed):
        return "{} runs {}".format(self.name, speed)

# Child class (inherits from Dog class)
class Bulldog(Dog):
    def run(self, speed):
        return "{} runs {}".format(self.name, speed)

# Dogs
p1 = Dog("Tom",6,True)
p2 = Dog("Fletcher",7,True)
p3 = Dog("Larry",9,True)

def dogs_status():
    if all(not i.is_hungry for i in Dog.all_dogs):
        mes = "My dogs are not hungry."
    else:
        mes = "My dogs are hungry."
    
    print(f"I have {Dog.cont} dogs. {p1.description()} {p2.description()} {p3.description()} And they are all {Dog.species}s, of course. {mes}")

dogs_status()
for i in Dog.all_dogs: i.eat()
dogs_status()
````
